a. Establish a linear regression model to predict y data (you can use sklearn).

```python

import numpy as np
```

```python
x_data = np.linspace(-0.5, 0.5, 200)[:, np.newaxis]
noise = np.random.normal(0, 0.02, x_data.shape)
y_data = np.square(x_data) + noise

x_data
```

```python
y_data
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
lr.fit(x_data,y_data)
```

```python
y_data_pred=lr.predict(x_data)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x_data,y_data,color='green',label='actual data')
plt.plot(x_data,y_data_pred,color='blue',label='predicted data')
plt.xlabel('x_data')
plt.ylabel('y_data')
```

# **Establish a polynomial model with highest power of 2 to predict y data (you can use sklearn). **

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly=PolynomialFeatures(degree=2)
x_poly=poly.fit_transform(x_data)

lr.fit(x_poly,y_data)
y_pred_poly=lr.predict(x_poly)
```

```python
plt.scatter(x_data,y_data,color='blue',label='actual data')
plt.plot(x_data,y_pred_poly,color='red',label='predicted data')
plt.xlabel('x_data')
plt.ylabel('y_data')
```

Establish a three-layer (including input, 1 hidden (6 nodes) and output layer) neural network to
predict y data (keras or tensorflow is recommended). Split the dataset into training 80% and
 testing data 20% using the following code in Python:


```python
import numpy as np
import tensorflow as tf
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt


#x_data = np.linspace(-0.5, 0.5, 200)[:, np.newaxis]
#noise = np.random.normal(0, 0.02, x_data.shape)
#y_data = np.square(x_data) + noise


x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.2, random_state=0)


model = tf.keras.Sequential([
    tf.keras.layers.Dense(6, input_dim=1, activation='relu'),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer='adam', loss='mean_squared_error')

history = model.fit(x_train, y_train, epochs=100, batch_size=32, validation_data=(x_test, y_test))


y_pred = model.predict(x_test)


plt.scatter(x_data, y_data, label='Original Data')
plt.plot(x_test, y_pred, color='red', linewidth=2, label='Neural Network Predictions')
plt.legend()
plt.xlabel('X Data')
plt.ylabel('Y Data')
plt.title('Neural Network Regression Example')
plt.show()
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mse_linear=mean_squared_error(y_data,y_data_pred)
mse_poly = mean_squared_error(y_data, y_pred_poly)
y_pred_nn = model.predict(x_test)
mse_nn = mean_squared_error(y_test, y_pred_nn)
print(mse_linear)
print(mse_poly)
print(mse_nn)
```
